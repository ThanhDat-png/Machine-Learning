```python
import pandas as pd 
import numpy as np 
import random as rnd 
import matplotlib.pyplot as plt 
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn import metrics 
from sklearn.model_selection import train_test_split
from sklearn import tree
from sklearn import datasets
```

### Tải dữ liệu Wine và chia tập dữ liệu

```python
dataset = datasets.load_wine()
X = dataset.data
y = dataset.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25)
data_df = pd.DataFrame(X, columns=dataset.feature_names)
data_df
```

### Khởi tạo và huấn luyện Tree Desicion

```python
model = DecisionTreeClassifier(max_depth=4)
model.fit(X_train, y_train)
```

### Dự đoán, Đánh giá và Vẽ biểu đồ cây quyết dịnh

```python
# Predict and show report
y_pred = model.predict(X_test)
print(f"Classification report for DecisionTree:\n"
f"{metrics.classification_report(y_test, y_pred)}\n")

# Plot confusion matrix
plt.figure(figsize= (5, 5))
ax = plt.gca()
disp = metrics.ConfusionMatrixDisplay.from_estimator(model, X_test, y_test, ax=ax)
_ = disp.figure_.suptitle("Confusion Matrix", fontsize=15)

#Show Decision Tree structure
fig, axes = plt.subplots(nrows=1, ncols=1, figsize=(20 ,5), dpi=500)
tree.plot_tree(model,
                feature_names=dataset.feature_names,
                class_names=dataset.target_names,
                filled=True);
```

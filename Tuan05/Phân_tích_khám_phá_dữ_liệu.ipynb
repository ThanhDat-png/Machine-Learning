```python
import numpy as np 
import pandas as pd 
from scipy import stats

# nạp tệp .csv vào dataframe 
covid_data = pd.read_csv("owid-covid-data.csv")
covid_data = covid_data[['iso_code','continent','location','date','total_cases','new_cases']]

# Xem nhanh dữ liệu
covid_data.head(5)
covid_data.dtypes
covid_data.shape

# Tính giá trị trung bình (mean)
data_mean = np.mean(covid_data["new_cases"])

# Tính giá trị trung vị (median)
data_median = np.median(covid_data["new_cases"])

# Tính yếu vị (mode)
data_mode = stats.mode(covid_data["new_cases"])

# Compute the range of the data
data_range = np.max(covid_data["new_cases"]) - np.min(covid_data["new_cases"])

# Tính phương sai (variance)
data_variance = np.var(covid_data["new_cases"])

# Tính độ lệch chuẩn ( standard deviation)
data_sd = np.std(covid_data["new_cases"])

# Tính giá trị lớn nhất và nhỏ nhất
data_max = np.max(covid_data["new_cases"])
data_min = np.min(covid_data["new_cases"])

# Tính bách phân vị thứ 60(60th percentile)
data_percentile = np.percentile(covid_data["new_cases"], 60)

# Tính tứ phân vị (quartile 0.75)
data_quartile = np.quantile(covid_data["new_cases"], 0.75)

# Tính khoảng tứ phân vị (IQR)
data_IQR = stats.iqr(covid_data["new_cases"])
```

```python
print("Mean:", data_mean)
print("Median:", data_median)
print("Mode:", data_mode)
print("Variance:", data_variance)
print("Standard deviation:", data_sd)
print("Range:", data_range)
print("60th percentile:", data_percentile)
print("Quartile (Q3):", data_quartile)
print("IQR:", data_IQR)
```

```python
import pandas as pd 
import matplotlib.pyplot as plt

# nạp dữ liệu và tạo biến mới
houseprices_data = pd.read_csv("data/HousingPricesData.csv")
houseprices_data = houseprices_data[['Zip','Price','Area','Room']]
houseprices_data['PriceperSqm'] = houseprices_data['Price']
houseprices_data['Area'] # Đơn giá/m2

houseprices_sorted = houseprices_data.sort_values('Price', ascending=False) # sắp xếp giảm dần theo giá

# Case 1: Biểu đồ cột cơ bản
plt.figure(figsize=(12, 6))
x = houseprices_sorted['Zip'][0:10]
y = houseprices_sorted['Price'][0:10]
plt.bar(x,y)
plt.show()

# Case 2: Biểu đồ cột nâng cao ( Thêm tiêu đề, nhãn trục, kích thước phông chữ)
plt.figure(figsize=(12, 6))
plt.bar(x, y)
plt.title('Top 10 Areas with the highest house prices', fontsize=15)
plt.xlabel('Zip code', fontsize = 12)
plt.xticks(fontsize=10)
plt.ylabel('House prices in millions',fontsize=12)
plt.yticks(fontsize=10)

#Case 3 : Ghép biểu đồ con (Subplots)
fig, ax = plt.subplots(figsize=(40, 18))
x = houseprices_sorted['Zip'][0:10]
y = houseprices_sorted['Price'][0:10]
y1 = houseprices_sorted["PriceperSqm"][0:10]

plt.subplot(1, 2, 1)
plt.bar(x, y)
plt.xticks(fontsize=17)
plt.ylabel('House prices in millions', fontsize=25)
plt.yticks(fontsize=20)
plt.title('Top 10 Areas with the highest house prices', fontsize=25)

plt.subplot(1, 2, 2)
plt.bar(x, y1)
plt.xticks(fontsize=17)
plt.ylabel('House prices per sqm', fontsize=25)
plt.yticks(fontsize=20)
plt.title('Top 10 Areas with the highes house prices per sqm','fontsize=25')
plt.show()
```

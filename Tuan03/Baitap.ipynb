```python
import matplotlib.pyplot as plt

tuoi = [25, 30, 22, 40, 35, 29, 50, 27, 31]
luong = [8000, 5000, 15000, -2000, 7000, 8500, 30000, 8200, 7800]

plt.hist(tuoi)
plt.xlabel("Tuổi")
plt.ylabel("Tần số")
plt.title("Phân phối Tuổi")
plt.show()
```

```python
plt.hist(luong)
plt.xlabel("Lương")
plt.ylabel("Tần số")
plt.title("Phân phối Lương")
plt.show()
```
